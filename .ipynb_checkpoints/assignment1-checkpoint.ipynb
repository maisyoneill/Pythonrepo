```python
#Task 1

transactions = [
    {"id": 1, "product": "Laptop", "price": 1200, "quantity": 2, "region": "EU"},
    {"id": 2, "product": "Phone", "price": 800, "quantity": -1, "region": "US"},
    {"id": 3, "product": "Tablet", "price": 0, "quantity": 5, "region": "EU"},
    {"id": 4, "product": "Monitor", "price": 350, "quantity": 3, "region": "US"},
    {"id": 5, "product": "Laptop", "price": 1200, "quantity": 1, "region": "EU"},
    {"id": 6, "product": "Keyboard", "price": 75, "quantity": 10, "region": "APAC"},
]
```

```python
## part a
valid_transaction = []
for t in transactions:
        if t["price"] > 0 and t["quantity"] >= 0:
            valid_transaction.append(t)
print(valid_transaction)
```

```python
## part b
for t in valid_transaction:
    t["total"] = t["price"] * t["quantity"]

## part c
regional_revenue = {}

for t in valid_transaction:
    region = t["region"]
    if region not in regional_revenue:
        regional_revenue[region] = 0
    regional_revenue[region] += t["total"]

## part d
highest_region = None
max_revenue = 0

for region, revenue in regional_revenue.items():
        if revenue > max_revenue:
            max_revenue = revenue
            highest_region = region

print("The region with the highest total revenue is", highest_region, "with", max_revenue)
```

```python
# Task 2
from statistics import mean
sales = [1200, 980, 1540, 870, 2010, 1650, 1300, 1100, 950, 1780, 2200, 1430, 1600, 1900]

total = sum(sales)
average = mean(sales)

best_day = 0
max_sales=0
for i, value in enumerate(sales, start=1):
    if value > max_sales:
        max_sales = value
        best_day = i

worst_day = 0
min_sales= float('inf')
for i, value in enumerate(sales, start=1):
    if value < min_sales:
        min_sales = value
        worst_day = i

above_average_count = 0
for s in sales:
    if s > mean(sales):
        above_average_count += 1

summarise_sales = dict({1: total, 2: average, 3:best_day, 4:worst_day, 5:above_average_count})

print(summarise_sales)
```

```python
# Task 3
students = [
    {"name": "Anna", "grade": 85, "attendance": 0.92},
    {"name": "Erik", "grade": 58, "attendance": 0.75},
    {"name": "Mia", "grade": 91, "attendance": 0.98},
    {"name": "Leo", "grade": 44, "attendance": 0.60},
    {"name": "Sara", "grade": 72, "attendance": 0.88},
    {"name": "Tom", "grade": 67, "attendance": 0.70},
    {"name": "Nina", "grade": 95, "attendance": 0.99},
    {"name": "Olli", "grade": 50, "attendance": 0.65},
]
```

```python
## part d = 0
for s in students: 
    if s["grade"] > g:
        g = s["grade"]
        highest_grade = s
print("The student with the highest grade is", highest_grade["name"], "with", g)
```

```python
# Task 4
class Portfolio:
    def __init__(self, owner):
        self.owner = owner
        self.holdings = {} 

## part a
    def add_asset(self, ticker, shares, price_per_share):
        if ticker in self.holdings:
            self.holdings[ticker]["shares"] = shares
            self.holdings[ticker]["price_per_share"] = price_per_share
        else:
            self.holdings[ticker] = {
                "shares": shares, 
                "price_per_share": price_per_share
        }

## part b
    def total_value(self):
        total = 0
        for data in self.holdings.values():
            total += data["shares"] * data["price_per_share"]
        return total

## part c
    def most_valuable(self):
        max_value = 0
        top_ticker = None

        for ticker, data in self.holdings.items():
            value = data["shares"] * data["price_per_share"]
            if value > max_value:
                max_value = value
                top_ticker = ticker 
    
        return top_ticker 

## part d
    def  __str__(self):
        return f"{self.owner} ({len(self.holdings)} assets, total value: {self.total_value()})"

## creation of instance 
p = Portfolio("Ellie")

p.add_asset("TULIP", 50, 240)
p.add_asset("POPPY", 3, 2750)
p.add_asset("LILY", 13, 730)

## updating an existing asset
p.add_asset("POPPY", 5, 2500)

print(p)
print("The total value is", p.total_value())
print("The most valuable asset is", p.most_valuable())
```

```python
# Task 5
from datetime import datetime

orders = [
    ("2025-01-15", "Laptop"), ("2025-01-22", "Phone"),
    ("2025-02-03", "Laptop"), ("2025-02-14", "Tablet"),
    ("2025-03-01", "Phone"), ("2025-03-10", "Phone"),
    ("2025-03-22", "Laptop"), ("2025-04-05", "Monitor"),
]

dates = []

# parse strings into date object
for order in orders:
    date_str = order[0]
    date_obj = datetime.strptime(date_str, "%Y-%m-%d").date()
    dates.append(date_obj)

# finding the difference
earliest = min(dates)
latest = max(dates)

difference = (latest - earliest).days

print("The number of days between the earliest and latest order is", difference)
```

```python
# Task 6
import numpy as np

revenue = np.array([
    [210, 230, 190, 250],
    [220, 260, 200, 270],
    [250, 290, 230, 310]])

years = [2022, 2023, 2024]

## part a
annual_totals = np.sum(revenue, axis=1)
for year, total in zip(years, annual_totals):
    print(f"{year}: {total}")
```

```python
## part b
quaterly_mean = np.mean(revenue, axis=0)
for year, total in zip(years, quaterly_mean):
    print(f"{year}: {total}")
```

```python
## part c
growth = np.array([1.05, 1.03, 1.08, 1.02])
adjusted_2024 = revenue [2] *growth
print("The adjusted values are", adjusted_2024)
```

```python
##part d
exceeding_values = revenue[revenue > 250]
print("Values above 250:", exceeding_values)
```

```python
# Task 7
import pandas as pd

data = {
    "employee": ["Anna", "Erik", "Mia", "Leo", "Sara", "Tom"],
    "department": ["Finance", "Marketing", "Finance", "IT", "Marketing", "IT"],
    "salary": [4500, 4000, 4800, 5200, 3900, 5100],
    "years_exp": [5, 3, 7, 10, 2, 8],
    "performance_score": [4.2, 3.8, 4.5, 4.0, 3.5, 4.9]
}

df = pd.DataFrame(data)
```

```python
## part a
df["salaray_per_year_exp"] = (df["salary"] / df["years_exp"]).round()
```

```python
## part b
sorting = df.sort_values(['performance_score'], ascending=[False]).head(3)
print(sorting)
```

```python
## part c
high_finance_performance = df.loc[
    (df["department"] == "Finance") & (df["performance_score"] > 4.0)
    ]
print("The finance high performers are:", high_finance_performance)
```

```python
## part d
df = df.rename(columns={"years_exp": "experience_years"})
```

```python
## part e
first_four = df.iloc[:4, [2,4]]
print("The salary and performance of the first four employees:", first_four)
```

```python
# Task 8
import numpy as np
import pandas as pd

rng = np.random.default_rng(seed=123)
days = pd.date_range(start="2025-01-06", periods=28, freq="D")
visitors = rng.integers(200, 600, size=28)
conversion_rate = rng.uniform(0.02, 0.08, size=28).round(4)
avg_order_value = rng.normal(50, 10, size=28).round(2)

## part a 
df = pd.DataFrame({
    "date": days, 
    "visitors": visitors,
    "conversion_rate": conversion_rate,
    "avg_order_value": avg_order_value
})

df = df.set_index("date")

## part b
df["conversions"] = (df["visitors"] * df["conversion_rate"]).round()
df["daily_revenue"] = df["conversions"] * df["avg_order_value"]

print(df)
```
